## 3. Find the weights

Weights come from an attached dataset rather than a download, which is
the difference between starting in seconds and starting in two minutes.

This cell **stops with an error** if the dataset is missing, rather than
quietly downloading eleven gigabytes instead. A silent fallback is how
someone waits two minutes every run without realising the dataset was
never being used.

```python
from pathlib import Path

WEIGHTS_SEARCH_ROOTS = (
    Path("/kaggle/input/datasets/johaness14/flux-2-4b-klein-jax"),
    Path("/kaggle/input/flux-2-4b-klein-jax"),
)

REQUIRED_ENTRIES = ("params", "tokenizer", "manifest.json")


def locate_weights() -> Path:
    """Find the attached dataset, checking both layouts Kaggle mounts under."""
    for candidate in WEIGHTS_SEARCH_ROOTS:
        if all((candidate / entry).exists() for entry in REQUIRED_ENTRIES):
            return candidate

    attached = sorted(p.name for p in Path("/kaggle/input").glob("*")) or ["(none)"]
    raise FileNotFoundError(
        "The weights dataset is not attached, or does not have the expected "
        f"layout {REQUIRED_ENTRIES}.\n"
        f"Looked in: {[str(p) for p in WEIGHTS_SEARCH_ROOTS]}\n"
        f"Found under /kaggle/input: {attached}\n"
        "Open the sidebar, choose Input, then Add Input, and attach the dataset."
    )


WEIGHTS_DIRECTORY = locate_weights()
print(f"weights: {WEIGHTS_DIRECTORY}")

# The fast tokenizer needs the full pipeline definition. Without it
# everything still works, but through a much slower path, so it is worth
# knowing which one this run is on.
has_definition = (WEIGHTS_DIRECTORY / "tokenizer" / "tokenizer.json").is_file()
print("fast tokenizer:", "yes" if has_definition else "no, falling back to the slower path")
```

## 4. Seed the compilation cache

Optional, and skipped automatically when there is nothing to seed.

Compiling the model takes a few minutes the first time. If a prebuilt
cache is attached as a second dataset, copying it here skips that wait
entirely.

The copy matters: `/kaggle/input` is read-only, and XLA needs somewhere
writable to store programs it compiles later. Pointing the cache
directly at a dataset would let it read but never write, so anything not
already there would be recompiled on every single run.

A cache is tied to the exact program that produced it, which includes
the code, the JAX version, and the chip. If any of those changed the
entries simply will not match, XLA compiles as it would have anyway, and
the only cost is time. Cell 6 reports which happened.

```python
import shutil
from pathlib import Path

CACHE_DIRECTORY = Path("/kaggle/working/compilation_cache")
CACHE_SEARCH_ROOTS = (
    Path("/kaggle/input/datasets/johaness14/flux-2-4b-klein-jax-cache"),
    Path("/kaggle/input/flux-2-4b-klein-jax-cache"),
)

CACHE_DIRECTORY.mkdir(parents=True, exist_ok=True)

seed_directory = next((path for path in CACHE_SEARCH_ROOTS if path.is_dir()), None)

if seed_directory is None:
    existing = len(list(CACHE_DIRECTORY.glob("*-cache")))
    print(f"no prebuilt cache attached; working cache holds {existing} entries")
else:
    copied = 0
    for entry in seed_directory.rglob("*-cache"):
        destination = CACHE_DIRECTORY / entry.name
        if not destination.exists():
            shutil.copy2(entry, destination)
            copied += 1
    total = len(list(CACHE_DIRECTORY.glob("*-cache")))
    print(f"copied {copied} cache entries from {seed_directory.name}; {total} now available")
```

## 5. Configure

Nothing here needs changing.

```python
import logging
from pathlib import Path

from src.config import (
    CheckpointSourceConfig,
    ExecutionConfig,
    InferenceConfig,
    MemoryResidencyStrategy,
)
from src.utils import configure_logging

WORKING_DIRECTORY = Path("/kaggle/working")

logger = configure_logging(
    log_file_path=WORKING_DIRECTORY / "generation_log.txt",
    logger_name="flux2_klein",
)

inference_config = InferenceConfig(
    checkpoint_source=CheckpointSourceConfig(
        # Pointing at the attached dataset means nothing is downloaded.
        local_cache_directory=WEIGHTS_DIRECTORY,
    ),
    # AUTO picks from the visible device count. On a v5e-8 that is fully
    # resident: every component stays in accelerator memory.
    residency_strategy=MemoryResidencyStrategy.AUTO,
)

execution_config = ExecutionConfig(
    compilation_cache_directory=CACHE_DIRECTORY,
)

print("configured")
```

## Notes

**After a kernel restart**, run cell 1 again before anything else. It
does not re-clone or re-download; it only restores the working directory
and the import path, which a restart clears.

**Saving the cache for later.** After a successful run,
`/kaggle/working/compilation_cache` holds everything compiled. Download
it and attach it as a dataset to skip the compile wait on a fresh
session. Re-export it whenever the code changes, since entries are tied
to the exact program that produced them.

**Three resolutions are offered**, and that is deliberate. Past roughly
4300 image tokens the reference sampling schedule switches to a formula
derived for a fifty-step model, and this checkpoint takes four steps.

**The full log**, including a timing breakdown of every stage, is
written to `/kaggle/working/generation_log.txt` and survives the
notebook output being cleared.
